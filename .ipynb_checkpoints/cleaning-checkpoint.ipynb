```python
df["timestamp"] = pd.to_datetime(df["timestamp"])
```

```python
df["date"] = df["timestamp"].dt.strftime("%Y-%m-%d")
df["time"] = df["timestamp"].dt.strftime("%H:%M:%S")
```

```python
df.head()
```

```python
df.drop(columns=["timestamp"], inplace=True)
```

```python
df["transaction type"].value_counts(normalize=True) * 100
```

```python
df.groupby("merchant_category")["amount (INR)"].sum().sort_values(ascending=False)
```

```python
df.groupby("hour_of_day")["transaction id"].count()
```

```python
df.groupby("hour_of_day")["transaction id"].count().plot(kind="line")
```

```python
df.groupby("is_weekend")["transaction id"].count()
```

```python
df.groupby("sender_age_group")["amount (INR)"].mean()
```

```python
df.groupby("receiver_age_group")["fraud_flag"].mean()
```

```python
df.groupby("sender_bank")["transaction id"].count().sort_values(ascending=False)
```

```python
df.groupby("sender_bank")["fraud_flag"].mean()
```

```python
df.groupby("device_type")["fraud_flag"].mean()
```

```python
df.groupby("network_type")["fraud_flag"].mean()
```

```python
df["high_value_flag"] = df["amount (INR)"].apply(lambda x: 1 if x > 2000 else 0)
```

```python
df.groupby("high_value_flag")["fraud_flag"].mean()
```

```python
df["risk_score"] = (
    df["fraud_flag"] +
    df["high_value_flag"] +
    df["is_weekend"]
)
```

```python
df.groupby("merchant_category")["risk_score"].mean().round(5)*100
```

```python
df.to_csv("upi_fintech_final.csv", index=False)
```
